```python
import os
import sys
import subprocess

def convert_fsdp_to_hf(actor_dir):
    """
    使用 verl 框架的 model_merger 将 FSDP 权重转换为 HuggingFace 格式。
    """
    # 自动拼接输出目录到 actor_dir 下的 huggingface 文件夹
    target_dir = os.path.join(actor_dir, "huggingface")
    
    # 前置检查
    if not os.path.exists(actor_dir):
        print(f"❌ 错误: 找不到指定的 actor 目录: {actor_dir}")
        return
        
    # 创建输出文件夹
    os.makedirs(target_dir, exist_ok=True)
    
    print("=" * 55)
    print("⚙️ 开始将 FSDP 权重转换为 HuggingFace 格式...")
    print(f"📂 输入路径 (FSDP): {actor_dir}")
    print(f"💾 输出路径 (HF)  : {target_dir}")
    print("=" * 55)
    
    # 构建 verl 的官方合并命令
    command = [
        sys.executable, "-m", "verl.model_merger", "merge",
        "--backend", "fsdp",
        "--local_dir", actor_dir,
        "--target_dir", target_dir
    ]
    
    try:
        # 执行命令并实时打印输出
        result = subprocess.run(command, check=True, text=True)
        print("=" * 55)
        print(f"✅ 转换成功！")
        print(f"你可以直接用 vLLM 或 Transformers 加载: {target_dir}")
        print("=" * 55)
        
    except subprocess.CalledProcessError as e:
        print("=" * 55)
        print(f"❌ 转换过程中发生错误！")
        print(f"返回码: {e.returncode}")
        print("请检查你的 verl 环境是否正常，或者由于 OOM 导致合并失败。")
        print("=" * 55)
    except KeyboardInterrupt:
        print("\n⚠️ 转换被用户手动中断。")

# --- 在 Jupyter 中直接指定路径 ---
ACTOR_DIR = "/mnt/oss/zwl/checkpoints/qwen3_4b_grpo_mixed/global_step_42/actor"

# 执行转换
convert_fsdp_to_hf(ACTOR_DIR)
```
