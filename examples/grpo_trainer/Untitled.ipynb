```python
import pandas as pd
import os

def shrink_parquet_file(file_path, shrink_factor=50):
    """
    读取 Parquet 文件，将其缩小指定的倍数，并以 _new 为后缀保存。
    """
    if not os.path.exists(file_path):
        print(f"❌ 错误: 文件不存在 {file_path}")
        return

    print(f"📂 正在加载: {file_path}")
    df = pd.read_parquet(file_path)
    original_size = len(df)
    
    # 计算抽样比例
    fraction = 1.0 / shrink_factor
    
    # 随机抽样 (random_state 保证每次运行抽样的结果一致)
    df_shrunk = df.sample(frac=fraction, random_state=42).reset_index(drop=True)
    shrunk_size = len(df_shrunk)
    
    # 生成新文件名 (例如: xxx_train.parquet -> xxx_train_new.parquet)
    base_name, ext = os.path.splitext(file_path)
    new_file_path = f"{base_name}_new{ext}"
    
    # 保存为新的 parquet 文件
    df_shrunk.to_parquet(new_file_path, engine='pyarrow')
    
    print(f"✅ 成功! 原数据量: {original_size} 行 -> 新数据量: {shrunk_size} 行")
    print(f"💾 新文件已保存至: {new_file_path}\n")

if __name__ == "__main__":
    # 定义你要处理的文件路径
    TRAIN_FILE = "/mnt/data/zwl/verl/data/rl/qwen3_4b_grpo_mixed_train.parquet"
    VAL_FILE = "/mnt/data/zwl/verl/data/rl/qwen3_4b_grpo_mixed_val.parquet"
    
    # 缩小 50 倍
    shrink_parquet_file(TRAIN_FILE, shrink_factor=50)
    shrink_parquet_file(VAL_FILE, shrink_factor=50)
```
